np.concatenate glues arrays together along an existing direction, keeping the total number of dimensions unchanged.

np.stack joins arrays along a brand-new direction, increasing the dimension count by one (e.g., turning 1D inputs into a 2D output).

1D Array Limitation: You cannot concatenate 1D arrays vertically because 1D arrays only have one axis—np.concatenate will simply join them end-to-end into a longer 1D array.

# 36. Write a function that converts a batch of raw scores into probabilities that sum to 1 per row. 

```python
def softmax(batch_score):
    shifted_row = batch_score - np.max(batch_score, axis=1,keepdims=True)
    exp_value = np.exp(shifted_row)
    return exp_value/np.sum(exp_value,axis=1,keepdims=True)
```

```python
import numpy as np

a = np.random.uniform(-2.0, 2.0, size=(5, 5))
print(a)
result = softmax(a)

print(np.round(result, 3))
print(np.sum(result,axis=1))
```
