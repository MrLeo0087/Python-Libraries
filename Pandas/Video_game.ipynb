### Find the top 10 best-selling games globally using nlargest() on Global_Sales.

```python
df.nlargest(n = 10, columns='Global_Sales',keep='first')  #using nlargest()
```

```python
df.sort_values('Global_Sales').head(10) #using sort method
```

### 3. Group by Genre and calculate total Global_Sales per genre, sorted descending.

```python
df.groupby(by='Genre')['Global_Sales'].sum().sort_values(ascending=False)
```

### 4. Build a pivot_table with Platform as rows, Genre as columns, and the mean Global_Sales as values.

```python
pivot_df = pd.pivot_table(df,index='Platform',columns=['Genre'],values='Global_Sales',aggfunc='mean')
```
